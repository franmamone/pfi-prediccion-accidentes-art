```python
from pathlib import Path
import sys

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))

from pfi_prediccion.config import ACCIDENTS_PARQUET, LEAKAGE_COLUMNS
from pfi_prediccion.eda import run_eda

df = pd.read_parquet(ACCIDENTS_PARQUET)
df["fecha_hora_accidente"] = pd.to_datetime(df["fecha_hora_accidente"])
print(f"Filas: {len(df):,} | Columnas: {df.shape[1]}")
df.head()
```

```python
nulls = df.isna().mean().sort_values(ascending=False).head(20) * 100
nulls.to_frame("pct_nulos").style.background_gradient(cmap="OrRd")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df["sexo_trabajador"].value_counts().plot(kind="bar", ax=axes[0], title="Sexo")
df["gravedad"].fillna("NULL").value_counts().plot(kind="bar", ax=axes[1], title="Gravedad")
plt.tight_layout()
```
